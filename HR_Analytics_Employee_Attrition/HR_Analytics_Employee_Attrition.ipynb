* **Job level** shows differences in attrition rates, which can help identify whether employees at particular career stages are more likely to leave.
* The analysis of **overtime and attrition** can help determine whether employees working overtime experience higher turnover, potentially indicating workload or work-life-balance concerns.
* Employees were grouped by **distance from home** to examine whether commuting distance is associated with higher attrition.

* **Job role shows substantial differences in retention.** Retention rates ranged from **97.50% for Research Directors** to **60.24% for Sales Representatives**, indicating that retention patterns vary considerably across job roles.

* **Employee tenure and workplace stability are associated with higher retention.** Employees who stayed had higher average:

  * **Years in Current Role:** 4.48 vs. 2.90 years
  * **Years with Current Manager:** 4.37 vs. 2.85 years

### 💡 Business Takeaway

The strongest retention patterns identified in this analysis are **overtime, job role, and employee tenure/stability**. Employees who do not work overtime, belong to certain job roles, and have longer tenure with the company, current role, and manager show higher retention rates.

These findings can help HR focus on **workload management, role-specific retention strategies, career development, and manager stability** to improve employee retention.

> **Note:** These findings represent associations observed in the dataset and **do not prove that these factors directly cause employees to stay or leave**.

```python
#Task 11 — HR Analytics Visualization

import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
# Count employees by attrition status
attrition_count = df["Attrition"].value_counts()

# Create bar chart
plt.bar(attrition_count.index, attrition_count.values,color="red")

# Add title and labels
plt.title("Employee Attrition")
plt.xlabel("Attrition Status")
plt.ylabel("Number of Employees")

# Add data labels
for i, value in enumerate(attrition_count.values):
    plt.text(i, value, str(value), ha="center", va="bottom")

plt.show()
```
