```python
#Task 8 — Customer Segmentation
# Check the customer summary created in Task 7
customer_summary.head()
```

```python
# 1. CREATE CUSTOMER SEGMENTS

# Create a function to classify customers based on their number of orders

def classify_customer(orders):
    # Customer with only 1 order
    if orders == 1:
        return "New"
    
    # Customer with 2 to 4 orders
    elif orders <= 4:
        return "Returning"
    
    # Customer with 5 or more orders
    else:
        return "Loyal"


# Apply the function to Number_of_Orders
customer_summary["Customer_Segment"] = (customer_summary["Number_of_Orders"].apply(classify_customer))

# Display the result
customer_summary.head(10)
```

```python
# 2. NUMBER OF CUSTOMERS IN EACH SEGMENT

segment_customers = (customer_summary["Customer_Segment"].value_counts().reset_index())

# Rename columns
segment_customers.columns = ["Customer_Segment","Number_of_Customers"]

# Display result
segment_customers
```

```python
# 3. REVENUE BY CUSTOMER SEGMENT

segment_revenue = (customer_summary.groupby("Customer_Segment")["Total_Spending"].sum()
    .reset_index().sort_values("Total_Spending", ascending=False))

# Display result
segment_revenue
```

```python
# 4. AVERAGE ORDER VALUE BY SEGMENT

segment_aov = (customer_summary.groupby("Customer_Segment")["Average_Order_Value"].mean()
    .reset_index().sort_values("Average_Order_Value", ascending=False))

# Display result
segment_aov
```

```python
# 5. AVERAGE ORDERS PER CUSTOMER

segment_orders = (customer_summary.groupby("Customer_Segment")["Number_of_Orders"].mean()
    .reset_index().sort_values("Number_of_Orders", ascending=False))

# Display result
segment_orders
```

```python
# 6. COMPLETE CUSTOMER SEGMENT SUMMARY

segment_summary = (customer_summary.groupby("Customer_Segment").agg(
        Number_of_Customers=("Customer ID", "nunique"),Total_Revenue=("Total_Spending", "sum"),
        Average_Spending=("Total_Spending", "mean"),Average_Order_Value=("Average_Order_Value", "mean"),
        Average_Orders=("Number_of_Orders", "mean")).reset_index())

# Sort by revenue
segment_summary = segment_summary.sort_values("Total_Revenue",ascending=False)

# Round numerical values
segment_summary = segment_summary.round(2)

# Display result
display(segment_summary)
```

```python
# 7. REVENUE CONTRIBUTION %

# Calculate the percentage of total revenue contributed by each customer segment

total_segment_revenue = segment_summary["Total_Revenue"].sum()

segment_summary["Revenue_Contribution_%"] = (segment_summary["Total_Revenue"]
                    / total_segment_revenue) * 100

# Round the percentage
segment_summary["Revenue_Contribution_%"] = (segment_summary["Revenue_Contribution_%"].round(2))

display(segment_summary)
```

Customers are segmented based on their number of orders to understand differences in purchasing behavior. Customers with one order are classified as New, customers with 2–4 orders as Returning, and customers with 5 or more orders as Loyal. The segments are compared based on customer count, revenue, average spending, average order value, and average number of orders.
