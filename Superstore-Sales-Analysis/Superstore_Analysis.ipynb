# PROJECT 1: SUPERSTORE SALES DATA CLEANING & TRANSFORMATION

```python
import pandas as pd

# Load the Superstore dataset
df = pd.read_csv(r"train.csv")
```

```python
# TASK 1: DATASET INSPECTION

# Display the first 10 rows to understand the dataset
df.head(10)
```

```python
# Check the number of rows and columns
df.shape
```

```python
# Display all column names
df.columns
```

```python
# Check the data type of each column
df.dtypes
```

```python
# Generate summary statistics for numerical columns
df.describe()
```

```python
# TASK 2: HANDLE MISSING VALUES

# Check the number of missing values in each column
df.isnull().sum()
```

```python
# Display rows where Postal Code is missing
df[df["Postal Code"].isnull()]
```

```python
# Check whether missing postal codes can be inferred from existing City and State information
df[(df["City"] == "Burlington") & (df["State"] == "Vermont")]
```

Postal Code values could not be reliably inferred from the available City/State information, so the missing
values are retained as NaN rather than creating incorrect data.

```python
# TASK 3: HANDLE DUPLICATE DATA

# Check for duplicate rows
df.duplicated().sum()
```

```python
# TASK 4: DATE CLEANING & TRANSFORMATION

# Convert Order Date and Ship Date into datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
```

```python
# Verify that both date columns have the correct data type
df[["Order Date", "Ship Date"]].dtypes
```

```python
# TASK 5: TEXT CLEANING

# Select columns containing text data
text_cols = ["Customer Name", "Ship Mode", "Segment", "Country","City", "State", 
             "Region", "Category","Sub-Category", "Product Name"]
```

```python
# Remove leading and trailing spaces from text values
df[text_cols] = df[text_cols].apply(lambda x: x.str.strip())
```

```python
# Standardize capitalization for names, locations and product names
df["Customer Name"] = df["Customer Name"].str.title()
df["City"] = df["City"].str.title()
df["State"] = df["State"].str.title()
df["Product Name"] = df["Product Name"].str.title()
```

```python
# Standardize capitalization of categorical values
df["Ship Mode"] = df["Ship Mode"].str.title()
df["Segment"] = df["Segment"].str.title()
df["Country"] = df["Country"].str.title()
df["Region"] = df["Region"].str.title()
df["Category"] = df["Category"].str.title()
df["Sub-Category"] = df["Sub-Category"].str.title()
```

```python
# Check unique values to verify categorical consistency
df["Ship Mode"].unique()
df["Segment"].unique()
df["Region"].unique()
df["Category"].unique()
df["Sub-Category"].unique()
```

```python
# TASK 6: NUMERICAL DATA CLEANING

# Check the data types of numerical columns
df[["Row ID", "Postal Code", "Sales"]].dtypes
```

```python
# Check summary statistics for numerical columns
df[["Row ID", "Postal Code", "Sales"]].describe()
```

```python
# Convert Postal Code to a nullable integer type so that integer values and missing values can coexist
df["Postal Code"] = df["Postal Code"].astype("Int64")
```

```python
# Verify the updated data type
df[["Row ID", "Postal Code", "Sales"]].dtypes
```

```python
# TASK 7: DATA TRANSFORMATION

# Calculate the number of days between order and shipment
df["Shipping Duration"] = (df["Ship Date"] - df["Order Date"]).dt.days
```

```python
# Extract the year from the Order Date
df["Order Year"] = df["Order Date"].dt.year
```

```python
# Calculate the median sales value for creating sales categories
median_sales = df["Sales"].median()
```

```python
# Categorize orders as High or Low based on median sales
df["Sales Category"] = df["Sales"].apply(
    lambda x: "High" if x > median_sales else "Low")
```

```python
# Extract the month from the Order Date
df["Order Month"] = df["Order Date"].dt.month
```

```python
# Display the newly created columns
df[["Shipping Duration","Order Year","Sales Category","Order Month"]].head()
```

```python
# TASK 8: FINAL DATA QUALITY CHECK

# Check remaining missing values
df.isnull().sum()
```

```python
# Check remaining duplicate rows
df.duplicated().sum()
```

```python
# Verify data types after cleaning and transformation
df.dtypes
```

```python
# Check the newly created columns
df[["Shipping Duration","Order Year","Sales Category"]].describe(include="all")
```

```python
# Check the values in Sales Category
df["Sales Category"].unique()
```

```python
# Check the range of Shipping Duration
df["Shipping Duration"].min()
df["Shipping Duration"].max()
```

```python
# Inspect the final cleaned dataset
df.head()
```

```python
# Save the cleaned dataset as a new CSV file
df.to_csv("superstore_cleaned.csv", index=False)
```
